## Noise

```python
def add_gaussian_noise(image, mean=0, sigma=25):
    """Add gaussian noise to an image."""
    gauss = np.random.normal(mean, sigma, image.shape).astype(np.float32)
    noisy_image = image.astype(np.float32) + gauss
    noisy_image = np.clip(noisy_image, 0, 255).astype(np.uint8)
    return noisy_image, gauss 

def impulse_noise(image, prob=0.05):
    """Add impulse noise to an image."""
    noisy_image = np.copy(image)
    rand = np.random.rand(*image.shape)
    noise_mask = np.zeros_like(image, dtype=np.uint8)
    noisy_image[rand < prob] = 0
    noise_mask[rand < prob] = 0

    noisy_image[rand > 1 - prob] = 255
    noise_mask[rand > 1 - prob] = 255
    return noisy_image, noise_mask

def poisson_noise(image, lam=4):
    poisson = np.random.poisson(lam, size=image.shape).astype(np.float32)
    noisy_image = image.astype(np.float32) + poisson
    noisy_image = np.clip(noisy_image, 0, 255).astype(np.uint8)
    return noisy_image, poisson

def uniform_noise(image, mean=0, var=25):
    a = mean - np.sqrt(3 * var)
    b = mean + np.sqrt(3 * var)

    uniform = np.random.uniform(a, b, size=image.shape).astype(np.float32)

    noisy_image = image.astype(np.float32) + uniform
    noisy_image = np.clip(noisy_image, 0, 255).astype(np.uint8)
    return noisy_image, uniform
```

## Filters

```python
def median_filter(image, kernel_size):
    h, w = image.shape

    pad = kernel_size // 2
    padded_img = np.pad(image, pad, mode='edge')
    img_new = np.zeros((h, w), dtype=float)

    for i in range(h):
        for j in range(w):
            window = padded_img[i:i+kernel_size, j:j+kernel_size].flatten()
            new_pixel = np.median(window)
            img_new[i, j] = new_pixel
    img_new = np.clip(img_new, 0, 255).astype(np.uint8)
    return img_new

def max_filter(image, kernel_size):
    h, w = image.shape

    pad = kernel_size // 2
    padded_img = np.pad(image, pad, mode='edge')
    img_new = np.zeros((h, w), dtype=float)

    for i in range(h):
        for j in range(w):
            window = padded_img[i:i+kernel_size, j:j+kernel_size].flatten()
            new_pixel = np.max(window)
            img_new[i, j] = new_pixel
    img_new = np.clip(img_new, 0, 255).astype(np.uint8)
    return img_new

def min_filter(image, kernel_size):
    h, w = image.shape

    pad = kernel_size // 2
    padded_img = np.pad(image, pad, mode='edge')
    img_new = np.zeros((h, w), dtype=float)

    for i in range(h):
        for j in range(w):
            window = padded_img[i:i+kernel_size, j:j+kernel_size].flatten()
            new_pixel = np.min(window)
            img_new[i, j] = new_pixel
    img_new = np.clip(img_new, 0, 255).astype(np.uint8)
    return img_new

def weighted_median(image, kernel_size):
    if kernel_size == 3:
        weights = np.array([[1, 2, 1],
                            [2, 3, 2],
                            [1, 2, 1]])
    elif kernel_size == 5:
        weights = np.array([[1, 1, 2, 1, 1],
                            [1, 2, 3, 2, 1],
                            [2, 3, 4, 3, 2],
                            [1, 2, 3, 2, 1],
                            [1, 1, 2, 1, 1]])
    h, w = image.shape
    pad = kernel_size // 2
    padded_img = np.pad(image, pad, mode='edge')
    img_new = np.zeros((h, w), dtype=float)

    for i in range(h):
        for j in range(w):
            window = padded_img[i:i+kernel_size, j:j+kernel_size]
            new_window = np.repeat(window.flatten(), weights.flatten())
            new_pixel = np.median(new_window)
            img_new[i, j] = new_pixel
    img_new = np.clip(img_new, 0, 255).astype(np.uint8)
    return img_new

def midpoint_filter(image, kernel_size):
    h, w = image.shape

    pad = kernel_size // 2
    padded_img = np.pad(image, pad, mode='edge')
    img_new = np.zeros((h, w), dtype=float)

    for i in range(h):
        for j in range(w):
            window = padded_img[i:i+kernel_size, j:j+kernel_size].flatten()
            min = np.min(window)
            max = np.max(window)
            new_pixel = (min+max)/2.0
            img_new[i, j] = new_pixel
    img_new = np.clip(img_new, 0, 255).astype(np.uint8)
    return img_new

def alpha_trimmed_mean_filter(image, kernel_size,d):
    h, w = image.shape
    pad = kernel_size // 2
    padded_img = np.pad(image, pad, mode='edge')
    img_new = np.zeros((h, w), dtype=float)
    for i in range(h):
        for j in range(w):
            window = padded_img[i:i+kernel_size, j:j+kernel_size].flatten()
            sorted_window = np.sort(window)
            new_window = sorted_window[d//2 : len(sorted_window) - d//2]
            new_pixel = np.mean(new_window)
            img_new[i, j] = new_pixel
    img_new = np.clip(img_new, 0, 255).astype(np.uint8)
    return img_new
```

## Local Filter

```python
def local_filter(image, region, median, kernel_size, d=None):
    h, w = image.shape
    pad = kernel_size // 2
    padded_img = np.pad(image, pad, mode='edge')
    img_new = np.zeros((h, w), dtype=float)

    pixels = region.ravel()
    noise_variance = np.var(pixels, ddof=0)
    
    for i in range(1, h-1):
        for j in range(1, w-1):
            local_window = padded_img[i:i+kernel_size, j:j+kernel_size].flatten()
            if median == "median":
                local_median = np.mean(local_window)
            elif median == "weighted":
                if kernel_size == 3:
                    weights = np.array([[1, 2, 1],
                                        [2, 3, 2],
                                        [1, 2, 1]])
                elif kernel_size == 5:
                    weights = np.array([[1, 1, 2, 1, 1],
                                        [1, 2, 3, 2, 1],
                                        [2, 3, 4, 3, 2],
                                        [1, 2, 3, 2, 1],
                                        [1, 1, 2, 1, 1]])
                new_window = np.repeat(local_window.flatten(), weights.flatten())
                local_median = np.median(new_window)
            elif median == "max":
                local_median = np.max(local_window)
            elif median == "min":
                local_median = np.min(local_window)
            elif median == "midpoint":
                min = np.min(local_window)
                max = np.max(local_window)
                local_median = (min+max)/2
            elif median == "alpha_trimmed":
                sorted_window = np.sort(local_window)
                new_window = sorted_window[d//2 : len(sorted_window) - d//2]
                local_median = np.mean(new_window)
            local_variance = np.var(local_window)
            new_pixel=image[i,j]-(noise_variance/local_variance)*(image[i,j]-local_median)
            img_new[i,j] = new_pixel
    img_new = np.clip(img_new, 0, 255).astype(np.uint8)

    return img_new
```

## Sharpening

```python
def sharpen(image, filtered, alpha):
    image = image.astype(np.float32)
    filtered = filtered.astype(np.float32)

    details = image - filtered
    enhanced = image + (details * alpha)

    enhanced = np.clip(enhanced, 0, 255)

    return enhanced.astype(np.uint8)
```
